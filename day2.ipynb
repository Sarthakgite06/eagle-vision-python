Data Type - Q 1

```python
# integer
# num
num1 = 10
print(num1)
print(type(num1))
print(id(num1))
# float
num2 = 20.5
print(num2)
print(type(num2))
print(id(num2))
# string
name = "sarthak"
print(name)
print(type(name))
print(id(name))
# boolean
is_student = True   
print(is_student)
print(type(is_student))
print(id(is_student))
# list
student = ["sarthak", "omkar", "raj"] 
print(student)
print(type(student))
print(id(student))
# tuple
student_tuple = ("sarthak", "omkar", "raj")
print(student_tuple)
print(type(student_tuple))
print(id(student_tuple))
# set
student_set = {"sarthak", "omkar", "raj"}  
print(student_set)
print(type(student_set))
print(id(student_set))
```
