```python
String1 = "hello i am sarthak and i am doing my job"
String2 = "am"

if String2 in String1:
    for i in range(len(String1)):
        if String1[i:i+len(String2)] == String2:
            print("Index:", i)
            break
```

```python
listone = [1,2,3,4]
listtwo = [5,6,8,10]
listone.extend(listtwo)
print(listone)
if len(listone) % 2 == 1:
    middle = len(listone) // 2
    print(listone[middle])

else:
    middle = len(listone) // 2
    previous = middle + 1
    median = (listone[middle] + listone[previous]) / 2    

print("Middle number is :",median)   
```
