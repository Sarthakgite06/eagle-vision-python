1. Create a Student Management System using Class and Object in Python. What to Do 1. Create a class named Student.  2. Create a constructor __init__() to initialize:  o Student name  o Roll number  o Age  o Marks of 3 subjects  3. Create a display_details() method to display all student information.  4. Create a calculate_total() method to calculate the total marks.  5. Create a calculate_percentage() method to calculate the percentage.  6. Create a check_result() method:  o Student passes if marks in every subject are 35 or above.  o Otherwise, display FAIL.  7. Create an update_marks() method to update the marks of a selected subject. 

```python
class Student:

    def __init__(self):
        self.name = ""
        self.roll_no = 0
        self.age = 0
        self.marks1 = 0
        self.marks2 = 0
        self.marks3 = 0

    def menu(self):

        print("\n1. Enter Student Details")
        print("2. Display Student Details")
        print("3. Calculate Total")
        print("4. Calculate Percentage")
        print("5. Check Result")
        print("6. Update Marks")
        print("7. Exit")

        choice = int(input("Enter your choice: "))

        match choice:

            case 1:
                self.enter_details()

            case 2:
                self.display_details()

            case 3:
                self.calculate_total()

            case 4:
                self.calculate_percentage()

            case 5:
                self.check_result()

            case 6:
                self.update_marks()

            case 7:
                print("Thank you for using Student Management System")

            case _:
                print("Invalid choice")
                self.menu()

    def enter_details(self):

        self.name = input("Enter name of student: ")
        self.roll_no = int(input("Enter roll no: "))
        self.age = int(input("Enter age: "))

        self.marks1 = int(input("Enter Marks of Python: "))
        self.marks2 = int(input("Enter Marks of Java: "))
        self.marks3 = int(input("Enter Marks of SQL: "))

        print("Student details added successfully")

        self.menu()

    def display_details(self):
        print("Student Name:", self.name)
        print("Roll Number:", self.roll_no)
        print("Age:", self.age)
        print("Python Marks:", self.marks1)
        print("Java Marks:", self.marks2)
        print("SQL Marks:", self.marks3)

        self.menu()

    def calculate_total(self):

        total_mark = self.marks1 + self.marks2 + self.marks3

        print("\nTotal Marks:", total_mark)

        self.menu()

    def calculate_percentage(self):

        total_mark = self.marks1 + self.marks2 + self.marks3

        percentage = total_mark / 3

        print("\nPercentage:", percentage, "%")

        self.menu()

    def check_result(self):

        if self.marks1 >= 35 and self.marks2 >= 35 and self.marks3 >= 35:

            print("Result: PASS")

        else:

            print("Result: FAIL")

            if self.marks1 < 35:
                print("Failed in Python")

            if self.marks2 < 35:
                print("Failed in Java")

            if self.marks3 < 35:
                print("Failed in SQL")

        self.menu()

    def update_marks(self):

        print("\n1. Python")
        print("2. Java")
        print("3. SQL")

        subject = int(input("Enter subject number: "))
        new_marks = int(input("Enter new marks: "))

        match subject:

            case 1:
                self.marks1 = new_marks
                print("Python marks updated:", self.marks1)

            case 2:
                self.marks2 = new_marks
                print("Java marks updated:", self.marks2)

            case 3:
                self.marks3 = new_marks
                print("SQL marks updated:", self.marks3)

            case _:
                print("Invalid subject number")

        self.menu()


obj = Student()
obj.menu()
```

2. Create a Library Management System using Class and Object in Python. What to Do 1. Create a class named Book.  2. Create a constructor __init__() to initialize:  o Book name  o Book ID  o Author name  o Availability status  3. Create a display_book() method to display book details.  4. Create a issue_book() method:  o Check whether the book is available.  o If available, issue the book and change its status.  o If already issued, display an appropriate message.  5. Create a return_book() method:  o Return the issued book.  o Change its availability status back to available.  6. Create a check_availability() method to display whether the book is available or issued. 

```python
class Book:

    def __init__(self):
        self.book_name = ""
        self.book_id = 0
        self.author_name = ""
        self.availability = True

    def menu(self):

        print("\n1. Enter Book Details")
        print("2. Display Book Details")
        print("3. Issue Book")
        print("4. Return Book")
        print("5. Check Availability")
        print("6. Exit")

        choice = int(input("Enter your choice: "))

        match choice:

            case 1:
                self.enter_book()

            case 2:
                self.display_book()

            case 3:
                self.issue_book()

            case 4:
                self.return_book()

            case 5:
                self.check_availability()

            case 6:
                print("Thank you for using Library Management System")

            case _:
                print("Invalid choice")
                self.menu()
    def enter_book(self):

        self.book_name = input("Enter Book Name: ")
        self.book_id = int(input("Enter Book ID: "))
        self.author_name = input("Enter Author Name: ")

        self.availability = True

        print("Book details added successfully")

        self.menu()
    def display_book(self):
        print("Book Name:", self.book_name)
        print("Book ID:", self.book_id)
        print("Author Name:", self.author_name)

        if self.availability:
            print("Availability: Available")
        else:
            print("Availability: Issued")

        self.menu()
    def issue_book(self):

        if self.availability:

            self.availability = False

            print("Book issued successfully")

        else:

            print("Book is already issued")

        self.menu()
    def return_book(self):

        if self.availability:

            print("Book is already available")

        else:

            self.availability = True

            print("Book returned successfully")

        self.menu()
    def check_availability(self):

        if self.availability:

            print("Book Status: Available")

        else:

            print("Book Status: Issued")

        self.menu()
obj = Book()
obj.menu()
```

3. Create a simple Employee Management System using Class and Object in Python. What to Do 1. Create a class named Employee.  2. Create a constructor __init__() to initialize:  o Employee name  o Employee ID  o Department  o Basic salary  3. Create a display_details() method to display employee information.  4. Create a calculate_salary() method:  o Add a fixed bonus of ₹5,000.  o Calculate and display the final salary.  5. Create a check_salary() method:  o If salary is ₹30,000 or above, display "Good Salary".  o Otherwise, display "Average Salary".  6. Create a menu-driven program:  o 1 → Display Details  o 2 → Calculate Salary  o 3 → Check Salary  o 4 → Exit 

```python
class Employee:

    def __init__(self, name, id, department, salary):
        self.name = name
        self.id = id
        self.department = department
        self.salary = salary

    def display_details(self):
        print("\nName:", self.name)
        print("ID:", self.id)
        print("Department:", self.department)
        print("Basic Salary:", self.salary)

    def calculate_salary(self):
        final_salary = self.salary + 5000
        print("Final Salary:", final_salary)

    def check_salary(self):
        final_salary = self.salary + 5000

        if final_salary >= 30000:
            print("Good Salary")
        else:
            print("Average Salary")

name = input("Enter Name: ")
id = input("Enter ID: ")
department = input("Enter Department: ")
salary = int(input("Enter Salary: "))

obj1 = Employee(name, id, department, salary)

while True:
    print("\n1. Display Details")
    print("2. Calculate Salary")
    print("3. Check Salary")
    print("4. Exit")

    choice = int(input("Enter choice: "))

    match choice:

        case 1:
            obj1.display_details()

        case 2:
            obj1.calculate_salary()

        case 3:
            obj1.check_salary()

        case 4:
            print("Program Ended")
            break

        case _:
            print("Invalid Choice")
```
