1. Student Class – __str__() Create a Student class with name, roll_no, and marks. Implement __str__() so that printing the 

```python
class student:
    def __init__(self,name,roll_no,marks):
        self.name = name
        self.roll = roll_no
        self.marks = marks
    def __str__(self):
        return self.name + str(self.roll) + str(self.marks)

obj1 = student("sarthak",21,87)
print(obj1)    
    
```

2. Product Class – __eq__() Create a Product class with name and price. Implement __eq__() to check whether two products have the same price.

```python
class product:
    def __init__(self,name,price):
        self.name = name
        self.price = price
    def __eq__(self, value2):
        return self.name == value2.name and self.price == value2.price
obj1 = product("laptop",5000)
obj2 = product("laptop",5000)    
print(obj1 == obj2)
```

3. Shopping Cart – __len__() Create a ShoppingCart class containing a list of products. Implement __len__() so that len(cart) returns the number of products in the cart.

```python
class ShoppingCart:
    def __init__(self, products):
        self.products = products

    def __len__(self):
        return len(self.products)


cart = ShoppingCart(["Laptop", "Mouse", "Keyboard"])

print(len(cart))
```

4. Bank Account – __str__() Create a BankAccount class with account holder name, account number, and balance. Implement __str__() to display account information when the object is printed. 


```python
class BankAccount:
    def __init__(self, name, account_number, balance):
        self.name = name
        self.account_number = account_number
        self.balance = balance

    def __str__(self):
        return self.name +" "+ str(self.account_number)+" " + str(self.balance)


account = BankAccount("Sarthak", 12345, 50000)

print(account)
```

. Number Class – __add__() Create a Number class that stores one number. Implement __add__() so that two objects can be added 

```python
class Number:
    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num


num1 = Number(10)
num2 = Number(20)

print(num1 + num2)
```

6. Create a Number class and implement: __add__()  __sub__()  __mul__()  __truediv__()  __mod__() 

```python
class Number:
    def __init__(self, num):
        self.num = num

    def __add__(self, other):
        return self.num + other.num

    def __sub__(self, other):
        return self.num - other.num

    def __mul__(self, other):
        return self.num * other.num

    def __truediv__(self, other):
        return self.num / other.num

    def __mod__(self, other):
        return self.num % other.num


num1 = Number(10)
num2 = Number(3)

print("Addition:", num1 + num2)
print("Subtraction:", num1 - num2)
print("Multiplication:", num1 * num2)
print("Division:", num1 / num2)
print("Modulus:", num1 % num2)
```
