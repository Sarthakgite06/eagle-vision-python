1. Write a program to count the number of vowels and consonants in a given string. 

```python
string = input("Enter a string: ").lower()

v = 0
c = 0

for i in string:
    if i in "aeiou":
        v += 1
    elif i.isalpha():
        c += 1

print("Vowels =", v)
print("Consonants =", c)
```

2. Write a program to check whether a given string is a Palindrome. - - Input: madam Output : Palindrome 

```python
str = input("enter name").lower().strip()
temp = "".join(reversed(str))
if str == temp:
    print("palindrome")
else:
    print("not palindrome")    
```

3. Write a program to reverse a string without using slicing ([::-1]) or built-in functions. 

```python
string = "omkar"

rev = ""

for i in string:
    rev = i + rev 
# original rev +=i we change order
print("Reversed String:", rev)
```

5. Write a program to remove all duplicate characters from a string while maintaining their original order. - - Input : programming Output : progamin 

```python
str = "programming"
result = ""

for i in string:
    if i not in result:
        result += i

print("Output:", result)
```

6. Write a program to find the longest word in a sentence. - Input: Python is an amazing programming language  - Output: programming


```python
string = "Python is an amazing programming language"

words = string.split()

longest = ""

for i in words:
    if len(i) > len(longest):
        longest = i
        # If the current word is longer than the previous longest word, update longest.

print("Longest word:", longest)
```

7. Write a program to capitalize the first letter of every word without using the title() method. - - Input:    welcome to python Output:   Welcome To Python 

```python
string = "welcome to python"

words = string.split()

for i in words:
    print(i[0].upper() + i[1:], end=" ")
```

8. Write a program to count the total number of words in a sentence without using the split() method. - - Input:     Python is easy to learn Output:    5 

```python
string = "Python is easy to learn"

count = 1         
# take count initial 1 first space

for i in string:
    if i == " ":
        count += 1

print("Total words:", count)
# Python is easy to learn
#       ↑    ↑     ↑  ↑
#    Space  Space Space Space

# Number of spaces = 4

# Words = Spaces + 1
#       = 4 + 1
#       = 5
```
