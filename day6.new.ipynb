2. Write a program to count the number of even and odd numbers in a tuple. 

```python
t = (12, 45, 7, 89, 23, 5, 67)
even = 0
odd = 0
for i in t:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
print("count of even number in t = ",even)
print("count of odd number in t = ",odd)            
```

5. Write a program to remove duplicate elements from a tuple while preserving the original order. 

```python
t = (10, 20, 30, 20, 40, 10, 50, 30)

result = []

for i in t:
    if i not in result:
        result.append(i)

new_tuple = tuple(result)

print("Original Tuple:", t)
print("Tuple after removing duplicates:", new_tuple)
```

6. Write a program to separate all even and odd elements of a tuple into two different tuples. - - -  - - - Input: (10,15,20,25,30) Output: Even = (10,20,30) Odd = (15,25) 

```python
t = (10,15,20,25,30)
even = []
odd = []
for i in t:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)
t1 = tuple(even)
t2 = tuple(odd)       
print("count of even number in t = ",t1)
print("count of odd number in t = ",t2) 
```

9. Write a program to convert a nested tuple like: Input - ((1,2), (3,4), (5,6)) into: output -(1,2,3,4,5,6) 

```python
t = ((1, 2), (3, 4), (5, 6))

result = []

for i in t:
    for j in i:
        result.append(j)

new_tuple = tuple(result)
print("output  Tuple:", new_tuple)
```
