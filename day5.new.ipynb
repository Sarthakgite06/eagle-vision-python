1. Write a program to count the number of even and odd numbers in a list. 


```python
numbers = [10, 15, 22, 7, 8, 13, 20]

even = 0
odd = 0
# take initial count as 0

for i in numbers:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1

print("Even numbers:", even)
print("Odd numbers:", odd)
```

2. Write a program to create a new list containing only the positive numbers from a given list. - - -  - - Input: [-10, 20, -5, 40, 0, 15] Output: [20, 40, 15] 

```python
numbers = [-10,20,-5,40,0,15]
positive = []
# declare empty list 
for i in numbers:
    if i > 0:
        positive.append(i)
        # add value of i in positive list
print(positive)        
```

3. Write a program to remove all duplicate elements from a list while preserving the original order. - - -  - - Input: [10,20,30,20,40,10,50] Output: [10,20,30,40,50] 

```python
numbers = [10,20,30,20,40,10,50]
result = []

for i in numbers:
    if i not in result:
        result.append(i)

print("Output:", result)
```

6. Write a program to count the frequency of every element in a list. - Input: - -  - - - - [1,2,3,2,1,4,1] Output: 1 -> 3 2 -> 2 3 -> 1 

```python
numbers = [1, 2, 3, 2, 1, 4, 1]

for i in numbers:
    if numbers.count(i) > 0:
        print(i, "->", numbers.count(i))
        numbers.remove(i)
        # to avoid repeatation remove visited
```

8. Write a program to check whether a list is a Palindrome. - Input: - -  - - [1,2,3,2,1] Output: Palindrome 

```python
numbers = [1, 2, 3, 2, 1]

temp = list(reversed(numbers))
# this is use for numbers only

if numbers == temp:
    print("Palindrome")
else:
    print("Not Palindrome")
```

9. Write a program to find all common elements between two lists. - List1 = [10,20,30,40] - -  - - List2 = [20,40,60] Output: [20,40] 

```python
List1 = [10,20,30,40]
List2 = [20,40,60]

common = []

for i in List1:
    if i in List2:
        common.append(i)
        # check and add in empty list

print(common)
```

10. Write a program to separate all even and odd numbers into two different lists. - - -  - - - [10,15,20,25,30] Output: Even = [10,20,30] Odd = [15,25] 

```python
numbers = [10, 15, 20, 25, 30]

even = []
odd = []

for i in numbers:
    if i % 2 == 0:
        even.append(i)
    else:
        odd.append(i)

print("Even =", even)
print("Odd =", odd)
```
