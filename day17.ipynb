Bank Account — Encapsulation Create a BankAccount class. • Create private variable __balance.  • Create deposit() and withdraw() methods.  • Create get_balance() getter.  • Don't allow withdrawal greater than balance.  • Don't allow negative deposit.  • Keep balance inaccessible directly from outside.  

```python
class BankAccount:
    def __init__(self, Account_no, balance):
        self.Account_no = Account_no
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            available_balance = self.get_balance()
            new_balance = available_balance + amount
            self.set_balance(new_balance)
            print("Amount deposited successfully")
        else:
            print("Enter a valid amount")

    def withdraw(self, amount):
        current_balance = self.get_balance()

        if amount > 0:
            if amount <= current_balance:
                new_balance = current_balance - amount
                self.set_balance(new_balance)
                print("Amount withdrawn successfully")
            else:
                print("Insufficient balance")
        else:
            print("Enter a valid amount")

    def display(self):
        print("Account no:", self.Account_no)
        print("Balance:", self.get_balance())

obj1 = BankAccount(77778888, 0)

obj1.deposit(2000)
obj1.withdraw(1000)
obj1.display()
```

10. Calculator — Method Overloading Create a Calculator class. Create an add() method that should support: add(10, 20) add(10, 20, 30) add(10, 20, 30, 40) • Use *args or default arguments.  • Do not create multiple add() methods.  • Similarly implement multiply() using *args.  

```python
class Calculator:

    def add(self, *args):
        total = 0

        for num in args:
            total = total + num

        return total

    def multiply(self, *args):
        total = 1

        for num in args:
            total = total * num

        return total
obj = Calculator()

print("Addition:", obj.add(10, 20))
print("Addition:", obj.add(10, 20, 30))
print("Addition:", obj.add(10, 20, 30, 40))

print("Multiplication:", obj.multiply(10, 20))
print("Multiplication:", obj.multiply(10, 20, 30))
print("Multiplication:", obj.multiply(10, 20, 30, 40))
```
