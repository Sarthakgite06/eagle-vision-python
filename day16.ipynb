7. Bank Account Create a BankAccount class with a private attribute __balance.  Create get_balance() to view the balance.  Create set_balance() to update the balance only if the amount is valid.  Create a withdraw() method with proper balance validation

```python
class Bank:
    def __init__(self):
        self.bank_details()

    def bank_details(self):
        self.account_no = 77778888
        self.__balance = 500000

    def get_balance(self):
        return self.__balance

    def set_balance(self, balance):
        self.__balance = balance


class Balance_Check(Bank):

    def withdraw(self, amount):
        current_balance = self.get_balance()

        if amount <= current_balance:
            new_balance = current_balance - amount
            self.set_balance(new_balance)
            print("Withdrawal successful")
        else:
            print("Insufficient balance")

    def display(self):
        print("Account No:", self.account_no)
        print("Balance:", self.get_balance())


obj1 = Balance_Check()

obj1.display()

obj1.withdraw(20000)

print("After withdrawal:")
obj1.display()
```

10. Multilevel Inheritance + Getter/Setter Person → private age with getter/setter.  Employee → salary.  Manager → department.  Use super() at each level 

```python
class Person:
    def __init__(self, age):
        self.__age = age

    def get_age(self):
        return self.__age

    def set_age(self, age):
        if age > 0:
            self.__age = age
            print("Age updated successfully")
        else:
            print("Invalid age")


class Employee(Person):
    def __init__(self, age, salary):
        super().__init__(age)
        self.salary = salary


class Manager(Employee):
    def __init__(self, age, salary, department):
        super().__init__(age, salary)
        self.department = department

    def display(self):
        print("Age:", self.get_age())
        print("Salary:", self.salary)
        print("Department:", self.department)
obj = Manager(30, 50000, "IT")
obj.display()
obj.set_age(35)
print("Updated Age:", obj.get_age())
```
