2. Write a function that accepts a list of numbers and returns the largest and smallest number.  

```python
def find_numbers(list1):
    largest = list1[0]
    smallest = list1[0]

    for i in list1:
        if i > largest:
            largest = i

        if i < smallest:
            smallest = i

    print("Largest number is:", largest)
    print("Smallest number is:", smallest)


find_numbers([10, 5, 20, 3, 15])
```

3. Write a function using  *args to calculate the sum, average, maximum, and minimum of any number of integers.  

```python
def calculate(*args):
    total = sum(args)
    average = total / len(args)
    maximum = max(args)
    minimum = min(args)

    print("Sum:", total)
    print("Average:", average)
    print("Maximum:", maximum)
    print("Minimum:", minimum)

calculate(10, 20, 30, 40, 50)
```

5. Write a function that accepts a string and returns the number of uppercase letters, lowercase letters, digits.  

```python
def String(str1):
    uppercase = 0
    lowercase = 0
    for char in str1:
        if char.isupper():
            uppercase += 1
        elif char.islower():
            lowercase += 1
    print("number of uppercase Letters:", uppercase)
    print("number of lowercase Letters:", lowercase)

String("Sarthak")
```

6. Write a function that checks whether a given string is a Palindrome. 

```python
def check_palindrome(Str1):
    reversed_str = Str1[::-1]
    if Str1 == reversed_str:
        print("The string is a palindrome.")
    else:
        print("The string is not a palindrome.")

check_palindrome("nitin")
```

7. Write a function that accepts a list of integers and returns a new list containing only even numbers.  


```python
def even(num):
    Even = []
    for i in num:
        if i%2 == 0:
            Even.append(i)
    print("Even numbers in the list:", Even)
even((1,2,3,4,5))    
```

8. Create a menu-driven calculator using functions that performs Addition, Subtraction, Multiplication, Division, and Exit.  

```python
def calculator(num1, num2, operation):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "multiply":
        return num1 * num2
    elif operation == "divide":
        if num2 != 0:
            return num1 / num2
        else:
            return "Error: Division by zero"
    else:
        return "Invalid operation"
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))
operation = input("Enter operation (add, subtract, multiply, divide): ")
result = calculator(num1, num2, operation)
print("Result:", result)
```

9. Write a function that accepts a dictionary of students and their marks, then returns the name of the topper and the highest marks.  


```python
def top(marks):
    highest_marks = max(marks.values())

    for name in marks:
        if marks[name] == highest_marks:
            print("Topper Name:", name)
            print("Highest Marks:", highest_marks)

top({"sarthak": 90, "om": 85, "jay": 95})
```

10. Write a function that accepts a sentence and returns the frequency of each word using a dictionary. 

```python
def frequency_count(string):
    frequency = {}
    for char in string:
        if char in frequency:
            frequency[char] += 1
        else:
            frequency[char] = 1
    return frequency
str1 = "sarthak"
print(frequency_count(str1))
```

11. Write a function that accepts a list of integers and returns:  • Sum of all numbers  • Count of even numbers  • Count of odd numbers  


```python
def check(a):
    even = 0
    odd = 0

    for i in a:
        if i % 2 == 0:
            even += 1
        else:
            odd += 1

    print("Sum:", sum(a))
    print("Even:", even)
    print("Odd:", odd)

check([1, 2, 3, 4, 5])
```
