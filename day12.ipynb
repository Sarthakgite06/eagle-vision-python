7. What will be the output? Explain the difference between the Local and Enclosing Namespace.  

```python
def outer():     
    x = 10      
    def inner():         
        x = 20         
        print(x)      
    inner()     
    print(x)  
outer() 
# Local: variable inside the current function.inner()
# Enclosing: variable inside the outer function.outer()
```

8. What will be the output? Why does nonlocal change the value of x?

```python
def outer():     
    x = 10      
    def inner():         
        nonlocal x         
        x = 50     
    inner()     
    print(x)  
outer() 
# 50 if x = 10 and inner function does nonlocal x; x = 50.
```
