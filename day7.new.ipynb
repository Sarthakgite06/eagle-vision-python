1. Write a program to find the union of two sets without using the union() method. 

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
set3 = set1.copy()
for i in set2:
    set3.add(i)
print("final union set: ",set3)    
```

2. Write a program to remove all duplicate elements from a list using a set. 

```python

list1 = [10, 20, 30, 20, 40, 10, 50, 30]

result = set(list1)
final_result = list(result)

print("List removing duplicates:", final_result)
```

3. Write a program to find all common elements between two lists using sets. 

```python
list1 = [12, 45, 7, 89, 23, 5, 67]
list2= [24,22,7,23,67]

Set1 = set(list1)
Set2 = set(list2)

Set4 = (Set1.intersection(Set2))

list3 = list(Set4)
print(list3)
```

4. Write a program to count the number of unique elements in a list using a set. 

```python
list1 = [10, 20, 30, 20, 40, 10, 50, 30]

unique = set(list1)
count = len(unique)
print("count of unique element: ",count)
```

5. Write a program to find the missing numbers between 1 to N using sets. - - - -  - - Input: N = 10 Numbers = {1,2,3,5,7,8,10} Output: {4,6,9} 

```python

numbers = {1, 2, 3, 5, 7, 8, 10}

missing = set(range(1, 11)) - numbers

print("Missing Numbers:", missing)
```

6. Write a program to create a new set containing only the even numbers from a given set. 

```python

set1 = {10, 15, 20, 25, 30, 35, 40}

even = set()

for i in set1:
    if i % 2 == 0:
        even.add(i)
print("Even Numbers:", even)
```

7. Write a program to merge three sets into a single set containing only unique elements. 

```python
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
set3 = {5, 6, 7, 8}

result = set1.copy()

for i in set2:
    result.add(i)

for i in set3:
    result.add(i)

print("Merged Set:", result)
```

8. Given two strings, find the common characters using sets.

```python
str1 = "apple"
str2 = "grape"

set1 = set(str1)
set2 = set(str2)

common = set1 & set2

print("Common Character:", common)
```

9. Given multiple lists, find all elements that appear in every list using sets. 


```python
list1 = [1, 2, 3, 4, 5]
list2 = [2, 3, 5, 6]
list3 = [3, 5, 7, 8]

set1 = set(list1)
set2 = set(list2)
set3 = set(list3)

common = set1 & set2 & set3

print("Common Elements:", common)
```
