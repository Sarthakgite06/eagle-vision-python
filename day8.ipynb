```python
# key(),value(),item()
dictone = {1:"user1",2:"user2",3:"user3"}
for i in dictone.items():
    print(i)

# access key
print(dictone.get(2)) 
# add
dictone[4] = "user4"   
print(dictone)
# update
dictone.update({5:"user5",6:"user6"})
print(dictone)
# pop(remove)
dictone.pop(2)
print(dictone)
# pop item(remove last value)
dictone.popitem()
print(dictone)
# clear it remove whole
dictone.clear()
print(dictone)
```

1. Write a program to count the frequency of each character in a string using a dictionary. - - -  - - Input: "programming" Output: {'p':1, 'r':2, 'o':1, ...} 

```python
Str1 = "programming"
freq = {}
for ch in Str1:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1
print(freq)        

```

2. Write a program to count the frequency of each element in a list using a dictionary. - - -  - - [10,20,10,30,20,10] Output: {10:3, 20:2, 30:1} 

```python
list1 = [10,20,10,30,20,10]
freq = {}
for i in list1:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
print(freq)            
```

3. Write a program to find the key having the maximum value in a dictionary. - - -  - - {'A':85, 'B':92, 'C':78} Output: B 

```python

dict1 = {'A': 85, 'B': 92, 'C': 78}

max_key = ""
# take key as empty and value as 0
max_value = 0

for key in dict1:
    if dict1[key] > max_value: 
        max_value = dict1[key]
        max_key = key

print("Key maximum value:", max_key)
```

6. Write a program to remove duplicate values from a dictionary. - - -  - - {'a':10, 'b':20, 'c':10, 'd':30} Output: {'a':10, 'b':20, 'd':30} 

```python
dict1 = {'a':10, 'b':20, 'c':10, 'd':30} 
result = {}
seen = set() 

for i,j in dict1.items(): 
    if j not in seen: 
        result[i]=j 
        seen.add(j)
print(result)


 
```

9. Write a program to create a dictionary from two lists: one containing keys and the other containing values. - - Keys = ['A', 'B', 'C'] Values = [10, 20, 30] 

```python
keys = ['A', 'B', 'C']
values = [10, 20, 30]
result = {}

# dict1 = dict(zip(keys, values))
# print(dict1)
for i in keys:
    for j in values:
     result[i] = j
print(result)
    
```

10. Write a program to create a student result dictionary containing name and marks of 5 students, then display the topper. 

```python
student = {}

for i in range(4):
    name = input("Enter student name: ")
    marks = int(input("Enter marks: "))
    student[name] = marks

topper = ""
highest = 0

for name in student:
    if student[name] > highest:
        highest = student[name]
        topper = name

print("Topper:", topper)
print("Marks:", highest)
```
