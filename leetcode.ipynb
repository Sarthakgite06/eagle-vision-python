```python
str1 = "abcdefghijkl"
result = []
if len(str1) % 4 == 0:
    for i in range(0,len(str1),4):
        word = str1[i:i+5]
        reverse_word = word[::-1]
        result.append(reverse_word)
else:
    print("The string is not divisible by 4")

print("".join(result))
```
