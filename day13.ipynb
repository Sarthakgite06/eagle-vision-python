2. Write a program to read the complete contents of a file and count the total number of characters.

```python
with open("temp1.txt", "r") as fobj:
    data = fobj.read()
    count = len(data)
    print(data)
    print("Number of characters:", count)
```

3. Write a program to read a file line by line and count the total number of lines.

```python
with open("temp1.txt", "r") as fobj:
    count = 0
    for i in fobj:
        count += 1
    print("Number of lines:", count)
```

4. Write a program to read a file and count the number of words present in it.

```python
with open("temp1.txt", "r") as fobj:
    data = fobj.read()
    words = data.split()

    count = 0
    for i in words:
        count += 1

    print("Number of words:", count)
```

8. Write a program to read a file and find the longest word present in it.  

```python
with open("temp2.txt", "r") as fobj:
    words = fobj.read().split()
    longest = ""
    for i in words:
        if len(i) > len(longest):
            longest = i
    print(longest)
```

9. Write a program to read a file containing numbers and calculate their sum, average, maximum, and minimum.  

```python
with open("temp3.txt", "r") as fobj:
    data = fobj.read().split()
    number = []
    for i in data:
        number.append(int(i))
    print(number)    
    print("sum of numbers:",sum(number))
    print("average of numbers:",sum(number)/len(number))
    print("maximum number:",max(number))
    print("minimum number:",min(number))
    
```
