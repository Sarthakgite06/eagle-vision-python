1. Student Information System Create a Student class with: • name  • roll_no  • branch  • marks  Create 3 student objects and create a method display() to print their details. 

```python
class Student:
    def __init__(self, name, roll_no, branch, marks):
        self.name = name
        self.roll_no = roll_no
        self.branch = branch
        self.marks = marks

    def display(self):
        print("Name:", self.name)
        print("Roll No:", self.roll_no)
        print("Branch:", self.branch)
        print("Marks:", self.marks)

obj1 = Student("Sarthak", 17, "AI & DS", 85)
obj2 = Student("Omkar", 10, "Computer", 78)
obj3 = Student("Jay", 3, "IT", 92)

obj1.display()
obj2.display()
obj3.display()
```

2. Employee Salary Create an Employee class with: • name  • employee_id  • salary  Create a method display_salary() that displays the employee's salary. Create 3 employee objects. : Don't assign the values directly inside the class. Pass them while creating objects. 

```python
class Employee:
    def __init__(self, name, employee_id, salary,location):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary
        self.location = location

    def display_salary(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Salary:", self.salary)
        print("location:",self.location)


# Creating 3 employee objects by passing values
obj1= Employee("ram", 101, 30000,"pune")
obj2 = Employee("om", 102, 35000,"mumbai")
obj3 = Employee("Raj", 103, 40000,"hydrabad")

# Display salary
obj1.display_salary()
obj2.display_salary()
obj3.display_salary()
```

3. Mobile Phone Create a Mobile class with: • brand  • model  • price  Create a method: display_details() Create 3 mobile objects with different information. Expected concept: Constructor + objects. 

```python
class Mobile:
    def __init__(self, brand, model, price):
        self.brand = brand
        self.model = model
        self.price = price

    def display_details(self):
        print("Brand:", self.brand)
        print("Model:", self.model)
        print("Price:", self.price)

obj1 = Mobile("Samsung", "Galaxy S24", 70000)
obj2 = Mobile("Apple", "iPhone 17", 65000)
obj3 = Mobile("Google", "pixal 9a", 45000)

obj1.display_details()
obj2.display_details()
obj3.display_details()
```
